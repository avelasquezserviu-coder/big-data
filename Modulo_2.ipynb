```python
# CSV
import pandas as pd

# df = pd.read_csv('proyectos.csv')

# Excel
df = pd.read_excel('/content/Base_Sintetica_SERVIU_DS27_1000_registros.xlsx')

# Comprobar carga
print(df.shape)
print(df.columns.tolist())
print(df.head())
```

```python
print(df.head())
print(df.shape)
print(df.info())
print(df.describe(include='all'))
print(df.isna().sum())
```

```python
df.isna().sum()
(df.isna().mean()*100).round(1)
```

```python
# Revisar filas idénticas
df.duplicated().sum()

# Revisar clave de negocio
df.duplicated(subset=['ID_Beneficiario']).sum()
```

```python
# Quitar espacios y unificar mayúsculas

df['Comuna'] = (
    df['Comuna'].astype('string')
      .str.strip()
      .str.title()
)

# Revisar categorías
print(df['Comuna'].value_counts(dropna=False))
```

```python
df['Fecha_Inicio_Obra'] = pd.to_datetime(
    df['Fecha_Inicio_Obra'], errors='coerce'
)
df['Fecha_Recepcion_Final_SERVIU'] = pd.to_datetime(
    df['Fecha_Recepcion_Final_SERVIU'], errors='coerce'
)

df['dias_desde_inicio'] = (
    df['Fecha_Recepcion_Final_SERVIU'] - df['Fecha_Inicio_Obra']
).dt.days
```

```python
# Brecha de duración en días (Real vs Planificado)

df['brecha_duracion_dias'] = (
    df['dias_desde_inicio'] -
    df['Duracion_Obra_Dias']
)

# Positiva: la obra tomó más días de lo planificado (atraso)
# Negativa o cero: la obra tomó los mismos o menos días de lo planificado

print(df[['Fecha_Inicio_Obra', 'Fecha_Recepcion_Final_SERVIU', 'dias_desde_inicio', 'Duracion_Obra_Dias', 'brecha_duracion_dias']].head())
```

```python
# Seleccionar columnas disponibles
sub = df[['Codigo_SERVIU', 'Comuna', 'Duracion_Obra_Dias', 'dias_desde_inicio', 'brecha_duracion_dias']]

# Filtrar obras con atraso (donde los días reales superaron la duración planificada)
en_atraso = df[df['brecha_duracion_dias'] > 0]

# Filtrar por comuna usando la columna correcta
maipu = df[df['Comuna'] == 'Maipú']

# Mostrar resultados preliminares
print(f"Proyectos en Maipú: {len(maipu)}")
print(f"Proyectos con atraso: {len(en_atraso)}")
display(sub.head())
```

```python
resumen = (
  df.groupby('Comuna', dropna=False)
    .agg(
      proyectos=('Codigo_SERVIU', 'nunique'),
      brecha_promedio_dias=('brecha_duracion_dias', 'mean')
    )
    .reset_index()
)
print(resumen)
```

```python
# 1. Separamos los datos en dos DataFrames para simular el escenario de integración
columnas_proyecto = ['Codigo_SERVIU', 'Region', 'Comuna', 'Tipo_Proyecto', 'Constructora']
proyectos = df[columnas_proyecto].drop_duplicates()

columnas_seguimiento = ['Codigo_SERVIU', 'Fecha_Inicio_Obra', 'Fecha_Recepcion_Final_SERVIU', 'Duracion_Obra_Dias', 'brecha_duracion_dias']
seguimiento = df[columnas_seguimiento].drop_duplicates()

# 2. Adaptamos el merge usando la columna llave correcta: 'Codigo_SERVIU'
integrado = proyectos.merge(
    seguimiento, on='Codigo_SERVIU', how='left'
)

# Mostramos las primeras filas del DataFrame integrado
display(integrado.head())
```
